# Openai Manager
Contains small snippets to access openai artifacts (i.e. files within vectorstore)

```python

# check files from vector store

from dotenv import load_dotenv
from openai import OpenAI
import os

load_dotenv(override=True)

api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("API-Key wurde geladen.")
    print("Anfang:", api_key[:12])
    print("Ende:", api_key[-4:])
    print("Länge:", len(api_key))
else:
    print("Kein OPENAI_API_KEY gefunden.")

client = OpenAI()

vector_store_id = "vs_6a21c391ff0481918eeecc10b21a1fdb"

files = client.vector_stores.files.list(
    vector_store_id=vector_store_id
)

for item in files.data:
    print(item.id, item.status)
```
